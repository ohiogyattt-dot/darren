```python
from google.colab import drive
drive.mount('/content/drive')

import sys
import os
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, HTML

# Folder proyek
FOLDER_PROYEK = "/content/drive/MyDrive/pbo_praktik"

# Agar Python dapat menemukan models.py
if FOLDER_PROYEK not in sys.path:
    sys.path.append(FOLDER_PROYEK)

# Import class dari file models.py
from models import Buku

FILE_CSV = os.path.join(FOLDER_PROYEK, "buku.csv")

def muat_data(b=None):
    output.clear_output()

    with output:
        if not os.path.exists(FILE_CSV):
            print("File buku.csv tidak ditemukan.")
            return

        # Membaca CSV
        df = pd.read_csv(FILE_CSV)

        # Membuat objek Buku
        daftar_buku = []

        for _, row in df.iterrows():
            buku = Buku(
                row["judul"],
                row["penulis"],
                row["isbn"],
                row["status"]
            )
            daftar_buku.append(buku)

        # Mengubah objek menjadi tabel
        data_tampil = []

        for buku in daftar_buku:
            data_tampil.append(buku.to_list())

        hasil = pd.DataFrame(
            data_tampil,
            columns=["Judul", "Penulis", "ISBN", "Status"]
        )

        display(hasil)

# Judul
display(HTML("""
<h2 style='color:#007acc'>
📚 Daftar Buku di Perpustakaan
</h2>
"""))

# Tombol refresh
btn_refresh = widgets.Button(
    description="Muat Ulang Data",
    button_style="primary",
    icon="refresh"
)

btn_refresh.on_click(muat_data)

output = widgets.Output()

display(btn_refresh)
display(output)

# Tampilkan data pertama kali
muat_data()
```
