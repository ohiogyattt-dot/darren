```python
from google.colab import drive
drive.mount("/content/drive")

import os
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output

# Lokasi file
file_path = "/content/drive/MyDrive/pbo_praktik/buku.csv"

# Area output
output = widgets.Output()


# Fungsi membersihkan dan menampilkan data
def baca_data():
    if not os.path.exists(file_path):
        return pd.DataFrame(
            columns=["judul", "penulis", "isbn", "status"]
        )

    data = pd.read_csv(file_path, dtype=str)

    data.columns = data.columns.str.strip().str.lower()
    data = data.T.groupby(level=0).first().T

    kolom = ["judul", "penulis", "isbn", "status"]

    for k in kolom:
        if k not in data.columns:
            data[k] = ""

    return data[kolom]


# Fungsi Tambah Buku
def buka_tambah_buku(b=None):
    with output:
        clear_output()

        display(
            widgets.HTML("<h3>Formulir Tambah Buku</h3>")
        )

        display(txt_judul)
        display(txt_penulis)
        display(txt_isbn)

        display(btn_simpan)


# Fungsi Tampilkan Buku
def buka_tampil_buku(b=None):
    with output:
        clear_output()

        display(
            widgets.HTML("<h3>Daftar Buku</h3>")
        )

        data = baca_data()

        if len(data) == 0:
            print("Belum ada data buku.")
        else:
            display(data)


# Fungsi Simpan Buku
def simpan_buku(b):
    judul = txt_judul.value.strip()
    penulis = txt_penulis.value.strip()
    isbn = txt_isbn.value.strip()

    with output:
        clear_output()

        if not judul or not penulis or not isbn:
            print("Semua kolom harus diisi!")

            display(txt_judul)
            display(txt_penulis)
            display(txt_isbn)
            display(btn_simpan)
            return

        data = baca_data()

        data_baru = pd.DataFrame(
            [[judul, penulis, isbn, "Tersedia"]],
            columns=["judul", "penulis", "isbn", "status"]
        )

        data = pd.concat(
            [data, data_baru],
            ignore_index=True
        )

        data.to_csv(file_path, index=False)

        print("✓ Buku berhasil ditambahkan")

        txt_judul.value = ""
        txt_penulis.value = ""
        txt_isbn.value = ""

        display(btn_tambah_lagi)
        display(btn_tampil_hasil)


# Fungsi Tambah Lagi
def tambah_lagi(b):
    buka_tambah_buku()


# Fungsi Tampilkan Hasil
def tampilkan_hasil(b):
    buka_tampil_buku()


# ===================================================
# GUI MENU UTAMA
# ===================================================

judul_menu = widgets.HTML("""
<h2 style='color:blue'>
Menu Utama Sistem Perpustakaan
</h2>
""")

btn_tambah = widgets.Button(
    description="Tambah Buku",
    button_style="success",
    layout=widgets.Layout(width="180px")
)

btn_tampil = widgets.Button(
    description="Tampilkan Buku",
    button_style="info",
    layout=widgets.Layout(width="180px")
)

# Form
txt_judul = widgets.Text(
    description="Judul:"
)

txt_penulis = widgets.Text(
    description="Penulis:"
)

txt_isbn = widgets.Text(
    description="ISBN:"
)

btn_simpan = widgets.Button(
    description="Simpan Buku",
    button_style="success"
)

# Tombol setelah simpan
btn_tambah_lagi = widgets.Button(
    description="Tambah Lagi",
    button_style="success"
)

btn_tampil_hasil = widgets.Button(
    description="Lihat Daftar Buku",
    button_style="info"
)

# Event
btn_tambah.on_click(buka_tambah_buku)
btn_tampil.on_click(buka_tampil_buku)
btn_simpan.on_click(simpan_buku)
btn_tambah_lagi.on_click(tambah_lagi)
btn_tampil_hasil.on_click(tampilkan_hasil)

# Tampilkan Menu
display(judul_menu)
display(btn_tambah)
display(btn_tampil)
display(output)
```
